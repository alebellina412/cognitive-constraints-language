# 7 — Learners against native speakers

If the crossover really marks the edge of an optimized kernel lexicon, then
speakers with more limited lexical resources should leave that regime earlier.
This notebook tests that on English produced by learners (COREFL) against native
English (SPGC).

Produces **Figure S4**. The length control the comparison depends on was
computed in notebook 3 (§3.4); this notebook reads its table back.

```python
import os
import subprocess
import sys

REPO = os.path.abspath("..") if os.path.isdir(os.path.join("..", "src")) else os.path.abspath(".")
sys.path.insert(0, os.path.join(REPO, "src"))

import plotting as P


def run(*command, must_succeed=True):
    """Run one pipeline step and, unlike a `!` cell, STOP if it fails.

    An IPython `!` cell throws away the exit status: a step that dies leaves no
    output, no error and no trace, and `nbconvert --execute` still reports the
    notebook as successful. Two defects in this pipeline's history hid exactly
    there, so every step below goes through this instead.

    `must_succeed=False` is used only for the two `--check` diagnostics of
    notebook 1, which print a loud banner rather than stopping the run.
    """
    print(">>", " ".join(str(c) for c in command), flush=True)
    code = subprocess.run([str(c) for c in command], check=False).returncode
    if code and must_succeed:
        raise RuntimeError(f"step failed with exit code {code} - read the output "
                           f"above; nothing after this point is valid")
    if code:
        rule = "*" * 72
        print(rule)
        print(f"*** THIS CHECK FAILED (exit code {code}). Read the output above")
        print("*** before going on: whatever depends on this corpus is missing")
        print("*** or wrong, and so is anything computed from it.")
        print(rule, flush=True)
    return code


def py(script, *args, must_succeed=True):
    """`run` for one of this repository's own scripts."""
    return run(sys.executable, os.path.join(REPO, "src", script), *args,
               must_succeed=must_succeed)


%matplotlib inline
USETEX = P.setup_style()
print("repo:", REPO, "| LaTeX text rendering:", USETEX)
```

## 7.1 Figure S4

Panel A is vocabulary growth, each curve averaged over random orderings of its
constituent texts with a 5–95% band — because D(t) of a corpus assembled from
many short texts depends on the order they are concatenated in, and any single
order is arbitrary.

Panel B plots R·p(R) in relative frequency, so a Zipfian regime is flat and the
end of the plateau is the crossover itself, marked at its fitted value rather
than eyeballed.

Three populations are tabulated but two are drawn. The COREFL native control is
genre-matched to the learners but at 52 k tokens is far too small to serve as the
native reference; it is reported for completeness.

The module asserts the conclusion at **matched length**: the learner kernel is
still the smaller one, with disjoint bands, once the length effect is removed.
